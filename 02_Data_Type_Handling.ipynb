## 2. Numerical Data

Numerical data represents quantities that support arithmetic operations such as sums, means, and comparisons.

Numerical data can be divided into two types:

- **Continuous Data:** Can take decimal values within a range.
- **Discrete Data:** Represents countable values, usually whole numbers.

In this dataset:

- `annual_income` and `purchase_amount` are continuous numerical features.
- `quantity` and `rating` are discrete numerical features.
- `age` is also numerical in nature, although it is currently stored as a string and will be converted later.

Identifying numerical data correctly is important because numerical operations and Machine Learning algorithms require appropriate numerical data types.

```python
# Identify numerical columns

numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numerical_columns.tolist())

print("\nSummary Statistics:")
df[numerical_columns].describe()
```

### Observation

The analysis also shows that some numerical features may require data type conversion before further preprocessing.

## 3. Categorical Data

Categorical data represents values drawn from a fixed, usually small, set of labels rather than measurable quantities.

In this dataset, the following columns are categorical:

- `gender`
- `city`
- `membership_type`
- `payment_method`

These columns are currently stored as string data, but they represent categories.

Identifying categorical columns correctly is important because categorical data is handled differently from numerical data during preprocessing and Machine Learning.

## 8. `astype()`

The `astype()` method is used to convert a Pandas column from one data type to another.

It is useful when we know that the values in a column are compatible with the target data type.

For example, an integer column can be converted to a float data type.

`astype()` performs direct type conversion. If the values cannot be converted to the specified type, it raises an error.

```python
# Example: Integer → Float

# Select an integer column from the dataset
integer_columns = df.select_dtypes(include=['int64', 'int32']).columns

if len(integer_columns) > 0:
    column = integer_columns[0]
    
    print("Column:", column)
    print("Data type before conversion:", df[column].dtype)

    df[column] = df[column].astype(float)

    print("Data type after conversion:", df[column].dtype)
else:
    print("No integer column found in the dataset.")
```

### Observation

The selected integer column was converted to the `float` data type using `astype()`.

This demonstrates how `astype()` can be used for direct data type conversion when the values are compatible with the target data type.

## 9. `to_numeric()`

The `to_numeric()` function is used to convert values into a numeric data type.

It is useful when numerical values are stored as strings.

The `errors` parameter can be used to handle values that cannot be converted:

- `errors='raise'` — raises an error for invalid values.
- `errors='coerce'` — converts invalid values to `NaN`.
- `errors='ignore'` — leaves invalid values unchanged.

This method is useful for safely converting columns that contain numerical values stored as text.

```python
# Example: Convert a column to numeric

numeric_column = 'children'

print("Data type before conversion:", df[numeric_column].dtype)

df[numeric_column] = pd.to_numeric(df[numeric_column], errors='coerce')

print("Data type after conversion:", df[numeric_column].dtype)

print("\nFirst 5 values:")
print(df[numeric_column].head())
```

### Observation

The `children` column was converted to a numeric data type using `to_numeric()`.

The `errors='coerce'` option converts values that cannot be interpreted as numbers into `NaN`, allowing the conversion to complete without an error.

## 10. `to_datetime()`

The `to_datetime()` function is used to convert values into a proper Date/Time data type.

It is useful when date values are stored as strings or objects.

Converting date columns to Date/Time format allows us to:

- Perform date calculations.
- Sort records by date.
- Extract year, month, and day.
- Calculate differences between dates.
- Perform time-based analysis.

The `errors` parameter can be used to handle invalid date values.

```python
# Example: String → Date/Time

date_column = 'reservation_status_date'

print("Data type before conversion:", df[date_column].dtype)

df[date_column] = pd.to_datetime(df[date_column], errors='coerce')

print("Data type after conversion:", df[date_column].dtype)

print("\nFirst 5 values:")
print(df[date_column].head())
```

### Observation

The `reservation_status_date` column was converted from its original data type to a proper Date/Time data type using `to_datetime()`.

The `errors='coerce'` option converts invalid date values into `NaT`, allowing the conversion to complete without stopping the process.

## 11. Handling Invalid Conversions

During data type conversion, some values may not be valid for the target data type.

For example, a column intended to contain numeric values may contain text or other invalid values.

Pandas provides the `errors` parameter to handle such values during conversion.

Common options are:

- `raise` — raises an error when conversion fails.
- `coerce` — converts invalid values to `NaN` for numeric conversion or `NaT` for date conversion.
- `ignore` — leaves the original values unchanged.

Handling invalid conversions prevents the preprocessing process from failing because of a small number of invalid values.

```python
# Demonstrate handling invalid numeric values

sample_values = pd.Series(['10', '25', 'invalid', '40'])

converted_values = pd.to_numeric(sample_values, errors='coerce')

print("Original Values:")
print(sample_values)

print("\nConverted Values:")
print(converted_values)
```

### Observation

The valid string values were converted into numeric values.

The invalid value was converted to `NaN` because `errors='coerce'` was used.

This allows invalid values to be identified and handled during later preprocessing steps.
