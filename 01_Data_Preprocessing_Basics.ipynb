## 10. Data Leakage

Data leakage occurs when information that should not be available to a Machine Learning model during training is unintentionally used by the model.

Data leakage can make a model appear to perform better than it actually would on unseen real-world data.

### Common Types of Data Leakage

- **Target Leakage:** Information related to the target becomes available to the model before prediction.
- **Train-Test Contamination:** Information from the test data influences the training process.
- **Feature Leakage:** A feature contains information that would not be available at prediction time.
- **Preprocessing Leakage:** Preprocessing is performed using information from the entire dataset before splitting the data.
- **Temporal Leakage:** Future information is used to predict an event that occurred earlier.

### Why is Data Leakage Important?

Data leakage can lead to:

- Overly optimistic model performance.
- Incorrect evaluation results.
- Poor performance on real-world unseen data.

To prevent leakage, preprocessing steps that learn information from the data should generally be fitted using the training data and then applied to validation and test data.

Data leakage will be considered throughout the Sprint 5 preprocessing workflow.
