# Module 7 - Encoding Categorical Variables

## 1. What is Categorical Encoding?

Categorical encoding is the process of converting categorical data into numerical values so that machine learning algorithms can understand and use the data.

Categorical variables contain labels or categories such as hotel type, meal type, customer type, or market segment.

Different encoding techniques are used depending on whether the categories are nominal, ordinal, binary, or high-cardinality.

```python
import pandas as pd
import numpy as np

# Load the raw dataset
df = pd.read_csv(r"C:\Users\HP\sprint-5-data-cleaning-preprocessing\data\hotel_bookings.csv")

# Display the first five records
df.head()
```

```python
categorical_columns = df.select_dtypes(include=['object', 'category']).columns

print("Categorical Columns:")
print(categorical_columns.tolist())

print("\nNumber of Categorical Columns:", len(categorical_columns))
```

### Observation

The categorical columns in the hotel bookings dataset were identified. These columns contain non-numerical values that may need to be encoded before being used in machine learning models.

## 2. Nominal Variables

Nominal variables are categorical variables whose categories do not have any natural order or ranking.

For example, in the hotel bookings dataset, `hotel`, `meal`, and `market_segment` are nominal variables because their categories represent different groups without any meaningful order.

Nominal variables are commonly encoded using One-Hot Encoding.

```python
nominal_columns = ['hotel', 'meal', 'market_segment']

print("Nominal Variables:")
for column in nominal_columns:
    print(f"\n{column}:")
    print(df[column].dropna().unique())
```

### Observation

The selected columns contain categories without a natural order. These nominal variables can be represented using encoding techniques such as One-Hot Encoding.

## 3. Ordinal Variables

Ordinal variables are categorical variables whose categories have a meaningful order or ranking.

For example, categories such as low, medium, and high have a natural order.

In the hotel bookings dataset, `customer_type` does not have a natural ranking, so it is not an ordinal variable. Therefore, an ordinal example is demonstrated using a separate ordered category.

Ordinal variables are commonly encoded using Ordinal Encoding when the category order is meaningful.

```python
ordinal_data = pd.Series([
    'Low',
    'Medium',
    'High',
    'Medium',
    'Low'
])

ordinal_order = {
    'Low': 1,
    'Medium': 2,
    'High': 3
}

encoded_data = ordinal_data.map(ordinal_order)

print("Original Values:")
print(ordinal_data.tolist())

print("\nEncoded Values:")
print(encoded_data.tolist())
```

### Observation

The ordered categories were converted into numerical values while preserving their natural ranking. Ordinal encoding should only be used when the categories have a meaningful order.

## 4. Binary Variables

Binary variables are categorical variables that have only two possible categories or outcomes.

Examples include Yes/No, True/False, or 0/1.

In the hotel bookings dataset, `is_canceled` and `is_repeated_guest` are binary variables because they contain two possible values: 0 and 1.

Binary variables can be represented directly using 0 and 1.

```python
binary_columns = ['is_canceled', 'is_repeated_guest']

print("Binary Variables:")

for column in binary_columns:
    print(f"\n{column}:")
    print(df[column].value_counts())
```

### Observation

The `is_canceled` and `is_repeated_guest` columns contain two possible values, 0 and 1. These binary variables are already represented numerically and can be directly used by machine learning algorithms.

## 5. Label Encoding

Label Encoding converts each category into a unique numerical value.

For example, categories such as A, B, and C can be converted into 0, 1, and 2.

Label Encoding is suitable for categorical variables where the numerical order does not create a misleading relationship, or when the encoding is used for appropriate tree-based models.

It should generally be avoided for nominal variables with models that may interpret the numbers as ordered values.

```python
from sklearn.preprocessing import LabelEncoder

df_label = df.copy()

column = 'hotel'

encoder = LabelEncoder()

df_label[column + '_encoded'] = encoder.fit_transform(df_label[column])

print("Original Values:")
print(df_label[column].unique())

print("\nEncoded Values:")
print(df_label[[column, column + '_encoded']].drop_duplicates())
```

### Observation

The `hotel` categories were converted into numerical labels using Label Encoding. The categories were assigned unique integer values, making the categorical data suitable for algorithms that can work with this type of representation.

## 6. Ordinal Encoding

Ordinal Encoding converts categorical values into numerical values while preserving the meaningful order of the categories.

For example:

Low → 1  
Medium → 2  
High → 3

Ordinal Encoding should be used only when the categories have a natural order or ranking.

It should not be used for nominal variables because assigning numbers can create a false sense of order.

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_data = pd.DataFrame({
    'priority': ['Low', 'Medium', 'High', 'Medium', 'Low']
})

encoder = OrdinalEncoder(
    categories=[['Low', 'Medium', 'High']]
)

ordinal_data['priority_encoded'] = encoder.fit_transform(
    ordinal_data[['priority']]
)

print(ordinal_data)
```

### Observation

The ordered categories were converted into numerical values while maintaining their natural ranking. Ordinal Encoding is appropriate when the order between categories has a meaningful interpretation.

## 7. One-Hot Encoding

One-Hot Encoding converts each category into a separate binary column.

For example, if a column contains three categories, each category gets its own column with values 0 or 1.

One-Hot Encoding is suitable for nominal variables where there is no natural order between categories.

It is useful for columns with a small or moderate number of unique categories.

```python
from sklearn.preprocessing import OneHotEncoder

column = 'meal'

encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown='ignore'
)

encoded_data = encoder.fit_transform(df[[column]])

encoded_columns = encoder.get_feature_names_out([column])

encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoded_columns,
    index=df.index
)

print("Original Column:")
print(df[[column]].head())

print("\nOne-Hot Encoded Data:")
print(encoded_df.head())
```

### Observation

The `meal` column was converted into separate binary columns using One-Hot Encoding. Each category is represented by 0 or 1 without introducing any artificial order between the categories.

## 8. Dummy Variables

Dummy Variables are binary variables created from categorical variables.

Each category is represented using 0 or 1. To avoid multicollinearity, one category can be dropped when using models that include an intercept.

For example, if a categorical column has three categories, two dummy variables can be created by dropping one category.

Dummy variables are commonly used for nominal categorical variables.

```python
column = 'meal'

dummy_data = pd.get_dummies(
    df[column],
    prefix=column,
    drop_first=True
)

print("Dummy Variables:")
print(dummy_data.head())

print("\nDummy Variable Columns:")
print(dummy_data.columns.tolist())
```

### Observation

Dummy variables were created from the `meal` column using 0 and 1 values. One category was dropped to reduce redundant information and avoid multicollinearity in suitable statistical or machine learning models.

## 9. Frequency Encoding

Frequency Encoding replaces each category with the number of times it appears in the dataset.

For example, if a category appears 1,000 times, it can be represented by 1,000 or by its frequency proportion.

Frequency Encoding is useful for categorical variables with many unique categories because it does not create a large number of new columns.

It should be used carefully because different categories can have the same frequency.

```python
column = 'market_segment'

frequency = df[column].value_counts()

df_frequency = df.copy()

df_frequency[column + '_frequency'] = df_frequency[column].map(frequency)

print("Original Values:")
print(df_frequency[[column, column + '_frequency']].head())

print("\nFrequency of Categories:")
print(frequency)
```

### Observation

Frequency Encoding was applied to the `market_segment` column. Each category was replaced with the number of times it occurs in the dataset, without creating additional columns for every category.

## 10. Target Encoding

Target Encoding replaces each category with a statistical value calculated from the target variable.

For example, each category can be replaced by the mean target value associated with that category.

Target Encoding can be useful for categorical variables with many unique categories, especially when One-Hot Encoding would create too many columns.

It should be applied carefully because it can cause target leakage if the target information is used incorrectly during encoding.

column = 'market_segment'
target = 'is_canceled'

target_mean = df.groupby(column)[target].mean()

df_target = df.copy()

df_target[column + '_target_encoded'] = df_target[column].map(target_mean)

print("Target Encoding:")
print(
    df_target[
        [column, target, column + '_target_encoded']
    ].head(10)
)

### Observation

Target Encoding was applied to the `market_segment` column using the mean cancellation rate for each category. This technique should be handled carefully to avoid target leakage, especially when preparing data for machine learning.

## 11. Handling Unknown Categories

Unknown categories are values that appear in new or unseen data but were not present when the encoding method was fitted.

This commonly occurs when a machine learning model receives new data after training.

Unknown categories should be handled carefully to prevent encoding errors. One approach is to use an encoder that ignores unknown categories and represents them without causing an error.

```python
from sklearn.preprocessing import OneHotEncoder

train_data = pd.DataFrame({
    'hotel': ['City Hotel', 'Resort Hotel', 'City Hotel']
})

test_data = pd.DataFrame({
    'hotel': ['City Hotel', 'New Hotel']
})

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

train_encoded = encoder.fit_transform(train_data[['hotel']])
test_encoded = encoder.transform(test_data[['hotel']])

print("Encoded Training Data:")
print(train_encoded)

print("\nEncoded New Data:")
print(test_encoded)
```

### Observation

The encoder was configured with `handle_unknown='ignore'`, allowing new categories to be processed without causing an error. This is useful when categories in new data are not present in the training data.

## 12. High-Cardinality Categories

High-cardinality categorical variables are columns that contain a large number of unique categories.

For example, a column containing many different customer IDs, booking IDs, or countries can have high cardinality.

Using One-Hot Encoding on high-cardinality columns can create a very large number of features and increase memory usage.

Frequency Encoding or Target Encoding can be considered for high-cardinality categories, depending on the dataset and the machine learning problem.

```python
categorical_columns = df.select_dtypes(include=['object', 'category']).columns

cardinality = df[categorical_columns].nunique().sort_values(ascending=False)

print("Number of unique categories:")
print(cardinality)

print("\nHigh-Cardinality Columns:")
print(cardinality[cardinality > 10])
```

### Observation

The number of unique categories in each categorical column was examined. Columns with many unique categories were identified as potential high-cardinality features, where One-Hot Encoding may create too many additional columns.

## 13. When to Use Each Encoding Technique

Different categorical encoding techniques are suitable for different types of categorical variables.

- Label Encoding: Useful for binary categories or suitable tree-based models.
- Ordinal Encoding: Use when categories have a meaningful order.
- One-Hot Encoding: Best for nominal variables with a small or moderate number of categories.
- Dummy Variables: Useful for nominal variables when one category should be dropped to reduce multicollinearity.
- Frequency Encoding: Useful for high-cardinality categorical variables.
- Target Encoding: Useful for high-cardinality variables when the target relationship is meaningful, but it must be applied carefully to avoid target leakage.

The encoding technique should be selected based on the type of categorical variable, number of categories, and machine learning algorithm.

```python
encoding_guide = pd.DataFrame({
    'Encoding Technique': [
        'Label Encoding',
        'Ordinal Encoding',
        'One-Hot Encoding',
        'Dummy Variables',
        'Frequency Encoding',
        'Target Encoding'
    ],
    'Best Used For': [
        'Binary or suitable tree-based models',
        'Ordered categories',
        'Nominal categories with fewer levels',
        'Nominal categories with multicollinearity concerns',
        'High-cardinality categories',
        'High-cardinality categories with a meaningful target relationship'
    ]
})

print(encoding_guide)
```

### Observation

Different encoding techniques are suitable for different categorical data. Choosing the appropriate technique helps represent categorical information correctly and reduces unnecessary features or misleading relationships.

## 14. Risks of Inappropriate Encoding

Choosing an inappropriate encoding technique can negatively affect data analysis and machine learning performance.

Common risks include:

- Creating a false order between nominal categories.
- Creating too many features when One-Hot Encoding is used for high-cardinality data.
- Increasing memory usage and model complexity.
- Causing target leakage when Target Encoding is applied incorrectly.
- Losing meaningful category relationships through inappropriate encoding.

Therefore, the encoding technique should be selected based on the type of categorical variable, its cardinality, and the machine learning algorithm being used.

```python
print("Categorical Encoding Risks:")

risks = [
    "False ordering between nominal categories",
    "Too many features with high-cardinality data",
    "Increased memory and model complexity",
    "Target leakage with improper target encoding",
    "Loss of meaningful category information"
]

for i, risk in enumerate(risks, 1):
    print(f"{i}. {risk}")
```

### Observation

Different encoding techniques have different risks. Using an inappropriate encoding method can introduce misleading relationships, increase the number of features, or cause data leakage. Therefore, encoding should be selected carefully based on the data and model requirements.
