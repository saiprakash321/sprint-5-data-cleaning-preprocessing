# Module 10 - Feature Selection

## 1. What is Feature Selection?

Feature Selection is the process of selecting the most relevant features from a dataset for use in analysis or machine learning models.

A dataset may contain useful, irrelevant, or redundant features. Feature selection helps identify the features that provide meaningful information for predicting the target variable.

Feature selection can reduce the number of features, simplify the model, improve computational efficiency, and help reduce the risk of overfitting.

```python
import pandas as pd
import numpy as np

# Load the raw dataset
df = pd.read_csv(r"C:\Users\HP\sprint-5-data-cleaning-preprocessing\data\hotel_bookings.csv")

# Display the first five records
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nFeatures in the Dataset:")
print(df.columns.tolist())

print("\nNumber of Features:", len(df.columns))
```

### Observation

The features available in the hotel bookings dataset were identified. Feature selection can be used to determine which features are useful for analysis and machine learning and which features can be removed.

## 2. Why Feature Selection?

Feature Selection is important because not every feature in a dataset contributes useful information to a machine learning model.

Selecting relevant features can:

- Reduce the number of input features.
- Reduce model complexity.
- Improve computational efficiency.
- Reduce the risk of overfitting.
- Improve model interpretability.
- Help the model focus on the most useful information.

Unnecessary features can add noise and may negatively affect the performance of machine learning models.

```python
print("Total Features:", df.shape[1])

print("\nFeature Names:")
for feature in df.columns:
    print("-", feature)
```

### Observation

The dataset contains multiple features that may not all be equally useful for machine learning. Feature selection helps identify important features and remove unnecessary information, which can simplify the model and improve efficiency.

## 3. Relevant Features

Relevant features are features that provide useful information for predicting or understanding the target variable.

For the hotel bookings dataset, features such as `lead_time`, `adr`, `adults`, `stays_in_week_nights`, and `total_of_special_requests` may provide useful information for predicting whether a booking is canceled.

Relevant features can improve model performance and help the model focus on meaningful patterns in the data.

```python
target = 'is_canceled'

candidate_features = [
    'lead_time',
    'adr',
    'adults',
    'stays_in_week_nights',
    'total_of_special_requests'
]

print("Target Variable:", target)

print("\nCandidate Relevant Features:")
for feature in candidate_features:
    print("-", feature)
```

### Observation

The selected numerical features are potential relevant features because they may contain useful information for predicting the target variable `is_canceled`. Their actual usefulness can be evaluated using feature selection techniques.

## 4. Irrelevant Features

Irrelevant features are features that do not provide useful information for predicting or understanding the target variable.

Including irrelevant features can add noise to the dataset and may make the machine learning model more complex.

For the hotel bookings dataset, features that do not have a meaningful relationship with the target variable may be considered irrelevant after proper analysis.

Irrelevant features should be removed only after evaluating their usefulness.

```python
target = 'is_canceled'

print("Target Variable:", target)

print("\nFeatures that require evaluation:")
for feature in df.columns:
    if feature != target:
        print("-", feature)
```

### Observation

All features other than the target variable were reviewed for feature selection. Features that provide little or no useful information can be considered irrelevant, but they should be removed only after appropriate analysis.

## 5. Redundant Features

Redundant features are features that provide similar or overlapping information.

When two or more features contain highly similar information, keeping all of them may add unnecessary complexity to the model.

For example, in the hotel bookings dataset, `stays_in_weekend_nights` and `stays_in_week_nights` both represent the length of a guest's stay and may contain related information.

Redundant features can be identified using correlation analysis and other feature selection techniques.

```python
numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'adr',
    'total_of_special_requests'
]

correlation_matrix = df[numeric_columns].corr()

print("Correlation Between Numerical Features:")
print(correlation_matrix.round(2))
```

### Observation

The correlation between numerical features was examined to identify features that may contain similar information. Highly correlated features may be redundant and can be considered for removal after evaluating their importance to the model.

## 6. Correlation-Based Selection

Correlation-Based Selection is a feature selection technique that uses the correlation between numerical features and the target variable.

A feature with a stronger correlation with the target may provide more useful information for prediction.

Highly correlated features with each other may also contain redundant information.

For the hotel bookings dataset, the correlation between numerical features and `is_canceled` can be examined to identify potentially useful features.

```python
target = 'is_canceled'

numeric_columns = df.select_dtypes(include=['number']).columns

correlation_with_target = (
    df[numeric_columns]
    .corr()[target]
    .drop(target)
    .sort_values(ascending=False)
)

print("Correlation with Target:")
print(correlation_with_target.round(3))
```

### Observation

The correlation between numerical features and the target variable `is_canceled` was calculated. Features with stronger positive or negative correlations may be more relevant to the target, while features with very weak correlation may require further evaluation before being selected or removed.

## 7. Variance Threshold

Variance Threshold is a feature selection technique that removes features with very low variance.

Variance measures how much the values of a feature differ from each other.

A feature with very low variance contains almost the same value for most records and may provide little useful information to a machine learning model.

Variance Threshold is mainly useful for numerical features.

```python
from sklearn.feature_selection import VarianceThreshold

numeric_columns = df.select_dtypes(include=['number']).columns

numeric_data = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

selector = VarianceThreshold(threshold=0.01)

selector.fit(numeric_data)

selected_features = numeric_data.columns[
    selector.get_support()
]

removed_features = numeric_data.columns[
    ~selector.get_support()
]

print("Selected Features:")
print(selected_features.tolist())

print("\nRemoved Low-Variance Features:")
print(removed_features.tolist())
```

### Observation

Variance Threshold was applied to the numerical features to identify features with very low variance. Features with variance below the selected threshold were considered less informative and removed from the selected feature set.

## 8. Univariate Feature Selection

Univariate Feature Selection evaluates each feature individually to determine how strongly it is related to the target variable.

For classification problems, statistical tests such as ANOVA F-test can be used to measure the relationship between numerical features and the target.

Features with higher scores are considered more informative for the target variable.

For the hotel bookings dataset, numerical features can be evaluated individually against `is_canceled`.

```python
from sklearn.feature_selection import SelectKBest, f_classif

target = 'is_canceled'

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'adr',
    'total_of_special_requests'
]

X = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

y = df[target]

selector = SelectKBest(
    score_func=f_classif,
    k='all'
)

selector.fit(X, y)

feature_scores = pd.DataFrame({
    'Feature': numeric_columns,
    'Score': selector.scores_
}).sort_values(
    by='Score',
    ascending=False
)

print("Univariate Feature Selection Scores:")
print(feature_scores.round(2))
```

### Observation

Each numerical feature was evaluated individually against the target variable `is_canceled`. Features with higher statistical scores show a stronger individual relationship with the target and may be more useful for prediction.

## 9. Mutual Information

Mutual Information is a feature selection technique that measures the amount of information a feature provides about the target variable.

A higher Mutual Information score indicates that a feature contains more useful information about the target.

Unlike correlation, Mutual Information can capture both linear and non-linear relationships between features and the target.

For the hotel bookings dataset, Mutual Information can be used to evaluate numerical features against `is_canceled`.

```python
from sklearn.feature_selection import mutual_info_classif

target = 'is_canceled'

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'adr',
    'total_of_special_requests'
]

X = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

y = df[target]

mi_scores = mutual_info_classif(
    X,
    y,
    random_state=42
)

mi_results = pd.DataFrame({
    'Feature': numeric_columns,
    'Mutual Information Score': mi_scores
}).sort_values(
    by='Mutual Information Score',
    ascending=False
)

print("Mutual Information Scores:")
print(mi_results.round(3))
```

### Observation

Mutual Information scores were calculated for the selected numerical features. Features with higher scores provide more information about the target variable and may be considered more useful for the machine learning model.

## 10. Recursive Feature Elimination

Recursive Feature Elimination (RFE) is a feature selection technique that repeatedly removes the least important features until the desired number of features remains.

RFE uses a machine learning model to evaluate feature importance and rank the features.

For the hotel bookings dataset, RFE can be used with a Logistic Regression model to select the most useful numerical features for predicting `is_canceled`.

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

target = 'is_canceled'

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'adr',
    'total_of_special_requests'
]

X = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

y = df[target]

model = LogisticRegression(
    max_iter=1000
)

selector = RFE(
    estimator=model,
    n_features_to_select=4
)

selector.fit(X, y)

rfe_results = pd.DataFrame({
    'Feature': numeric_columns,
    'Selected': selector.support_,
    'Ranking': selector.ranking_
}).sort_values(
    by='Ranking'
)

print("RFE Feature Selection Results:")
print(rfe_results)
```

### Observation

Recursive Feature Elimination ranked the selected numerical features based on their importance to the Logistic Regression model. The features with a ranking of 1 were selected as the most useful features.

## 11. Feature Importance

Feature Importance measures how useful each feature is for making predictions in a machine learning model.

Tree-based models such as Decision Trees and Random Forest can calculate feature importance based on how much each feature contributes to the model's decisions.

Features with higher importance scores contribute more to the model's predictions, while features with very low importance may be considered for removal after proper evaluation.

```python
from sklearn.ensemble import RandomForestClassifier

target = 'is_canceled'

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'adr',
    'total_of_special_requests'
]

X = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

y = df[target]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X, y)

feature_importance = pd.DataFrame({
    'Feature': numeric_columns,
    'Importance': model.feature_importances_
}).sort_values(
    by='Importance',
    ascending=False
)

print("Feature Importance:")
print(feature_importance.round(4))
```

### Observation

Feature importance was calculated using a Random Forest model. Features with higher importance scores contribute more to the model's predictions, while features with lower scores may be considered less important.
