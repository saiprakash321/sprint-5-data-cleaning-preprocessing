# Module 9 - Data Transformation

## 1. Why Transform Data?

Data transformation is the process of changing the scale, distribution, or structure of data into a more suitable form for analysis and machine learning.

Data transformation is useful for:

- Reducing the effect of skewed data.
- Reducing the influence of extreme values.
- Making numerical features more suitable for machine learning algorithms.
- Improving the distribution of numerical variables.
- Making different features easier to compare.

For the hotel bookings dataset, numerical features such as `lead_time` and `adr` may have skewed distributions. Transformation techniques can be used to make these features more suitable for analysis.

```python
import pandas as pd
import numpy as np

# Load the raw dataset
df = pd.read_csv(r"C:\Users\HP\sprint-5-data-cleaning-preprocessing\data\hotel_bookings.csv")

# Display the first five records
df.head()
```

```python
numeric_columns = ['lead_time', 'adr']

print("Summary Statistics Before Transformation:")

for column in numeric_columns:
    print(f"\n{column}:")
    print(df[column].describe())
```

### Observation

The selected numerical features were examined using summary statistics before applying any transformation. The distribution and range of these features can be analyzed to determine whether transformation is required.

## 2. Log Transformation

Log Transformation is a data transformation technique used to reduce right-skewness and decrease the influence of extreme values.

It compresses large values more than small values, which can make a highly skewed distribution more balanced.

Log Transformation is commonly applied to positive numerical features such as `lead_time` and `adr`.

For values that can contain zero, `log1p()` can be used because it calculates log(1 + x).

```python
import numpy as np

column = 'lead_time'

df_log = df.copy()

df_log[column + '_log'] = np.log1p(df_log[column])

print("Before Log Transformation:")
print(df[column].describe())

print("\nAfter Log Transformation:")
print(df_log[column + '_log'].describe())
```

### Observation

The `lead_time` feature was transformed using Log Transformation. The transformation compresses large values and can reduce the effect of extreme values and right-skewness in the distribution.

## 3. Square Root Transformation

Square Root Transformation is a data transformation technique used to reduce moderate right-skewness in numerical data.

It compresses larger values more than smaller values, but its effect is usually less strong than Log Transformation.

Square Root Transformation can be applied to non-negative numerical features such as `lead_time` in the hotel bookings dataset.

```python
import numpy as np

column = 'lead_time'

df_sqrt = df.copy()

df_sqrt[column + '_sqrt'] = np.sqrt(df_sqrt[column])

print("Before Square Root Transformation:")
print(df[column].describe())

print("\nAfter Square Root Transformation:")
print(df_sqrt[column + '_sqrt'].describe())
```

### Observation

The `lead_time` feature was transformed using Square Root Transformation. The transformation compresses larger values and can help reduce moderate right-skewness in the distribution.

## 4. Power Transformation

Power Transformation is a data transformation technique used to make numerical data more symmetric and reduce skewness.

It applies a mathematical power to the values of a feature.

Power transformations can be useful when numerical data has a skewed distribution and needs to be made more suitable for analysis or machine learning.

Common power transformations include Box-Cox and Yeo-Johnson transformations.

```python
import numpy as np

column = 'lead_time'

df_power = df.copy()

df_power[column + '_power'] = np.power(
    df_power[column],
    0.5
)

print("Before Power Transformation:")
print(df[column].describe())

print("\nAfter Power Transformation:")
print(df_power[column + '_power'].describe())
```

### Observation

The `lead_time` feature was transformed using a power of 0.5, which is equivalent to a square root transformation. The transformation reduces the effect of larger values and can help reduce skewness in the distribution.

## 5. Box-Cox Transformation

Box-Cox Transformation is a power transformation technique used to reduce skewness and make numerical data more normally distributed.

It can only be applied to strictly positive values.

For the hotel bookings dataset, `adr` can contain zero or negative values, so a positive shift is applied before using the Box-Cox transformation.

```python
from scipy.stats import boxcox

column = 'adr'

df_boxcox = df.copy()

shift = 1 - df_boxcox[column].min()
positive_values = df_boxcox[column] + shift

df_boxcox[column + '_boxcox'], lambda_value = boxcox(
    positive_values
)

print("Box-Cox Lambda:", lambda_value)

print("\nBefore Box-Cox Transformation:")
print(df[column].describe())

print("\nAfter Box-Cox Transformation:")
print(df_boxcox[column + '_boxcox'].describe())
```

### Observation

The `adr` feature was transformed using the Box-Cox method after shifting the values to make them strictly positive. The transformation can help reduce skewness and make the distribution more suitable for statistical analysis.

## 6. Yeo-Johnson Transformation

Yeo-Johnson Transformation is a power transformation technique used to reduce skewness and make numerical data more suitable for analysis and machine learning.

Unlike Box-Cox Transformation, Yeo-Johnson can be applied to data containing zero and negative values.

For the hotel bookings dataset, Yeo-Johnson can be applied directly to the `adr` feature because it can handle non-positive values.

```python
from sklearn.preprocessing import PowerTransformer

column = 'adr'

df_yeojohnson = df.copy()

transformer = PowerTransformer(method='yeo-johnson')

df_yeojohnson[column + '_yeojohnson'] = transformer.fit_transform(
    df_yeojohnson[[column]]
)

print("Before Yeo-Johnson Transformation:")
print(df[column].describe())

print("\nAfter Yeo-Johnson Transformation:")
print(df_yeojohnson[column + '_yeojohnson'].describe())
```

### Observation

The `adr` feature was transformed using the Yeo-Johnson method. Unlike Box-Cox, Yeo-Johnson can handle zero and negative values and can help reduce skewness in the distribution.

## 7. Handling Skewed Data

Skewed data is data whose distribution is not symmetric around its mean.

There are two common types of skewness:

- Right Skewness: The distribution has a longer tail toward higher values.
- Left Skewness: The distribution has a longer tail toward lower values.

Skewed numerical features can affect statistical analysis and some machine learning algorithms.

Transformation techniques such as Log, Square Root, Box-Cox, and Yeo-Johnson can be used to reduce skewness.

For the hotel bookings dataset, `lead_time` and `adr` can be examined for skewness.

```python
numeric_columns = ['lead_time', 'adr']

print("Skewness Before Transformation:")

for column in numeric_columns:
    print(f"{column}: {df[column].skew():.2f}")
```

### Observation

The skewness of the selected numerical features was calculated. Features with high positive or negative skewness may require transformation to obtain a more balanced distribution.

## 8. Normalization

Normalization is a data transformation technique that scales numerical features to a specific range, usually between 0 and 1.

It is useful when numerical features have different ranges and when the machine learning algorithm is sensitive to feature magnitude.

Min-Max Scaling is a commonly used normalization technique.

Normalization can help distance-based algorithms such as KNN and K-Means work more effectively.

```python
from sklearn.preprocessing import MinMaxScaler

numeric_columns = ['lead_time', 'adr']

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(
    df[numeric_columns]
)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=numeric_columns,
    index=df.index
)

print("Before Normalization:")
print(df[numeric_columns].head())

print("\nAfter Normalization:")
print(normalized_df.head())

print("\nMinimum Values:")
print(normalized_df.min())

print("\nMaximum Values:")
print(normalized_df.max())
```

### Observation

The selected numerical features were normalized to a range between 0 and 1. This brings the features to a common scale and reduces the effect of differences in their original ranges.

## 9. Standardization

Standardization is a data transformation technique that changes numerical features so that they have a mean of approximately 0 and a standard deviation of approximately 1.

The standardized value is calculated using:

Z = (X - Mean) / Standard Deviation

Standardization is useful when numerical features have different scales and is commonly used with algorithms such as Logistic Regression, SVM, KNN, and PCA.

Unlike normalization, standardization does not restrict values to a fixed range such as 0 to 1.

```python
from sklearn.preprocessing import StandardScaler

numeric_columns = ['lead_time', 'adr']

scaler = StandardScaler()

standardized_data = scaler.fit_transform(
    df[numeric_columns]
)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=numeric_columns,
    index=df.index
)

print("Before Standardization:")
print(df[numeric_columns].head())

print("\nAfter Standardization:")
print(standardized_df.head())

print("\nMean After Standardization:")
print(standardized_df.mean().round(2))

print("\nStandard Deviation After Standardization:")
print(standardized_df.std().round(2))
```

### Observation

The selected numerical features were standardized using `StandardScaler`. The transformed features have a mean close to 0 and a standard deviation close to 1.

## 10. Compare Distribution Before and After Transformation

Comparing the distribution before and after transformation helps determine whether a transformation has reduced skewness and improved the shape of the data.

The `lead_time` feature is used for comparison because it contains non-negative numerical values and can be transformed using Log Transformation.

The skewness value and summary statistics are compared before and after transformation.

```python
import numpy as np
import pandas as pd

column = 'lead_time'

original = df[column].dropna()
log_transformed = np.log1p(original)

comparison = pd.DataFrame({
    'Before Transformation': [
        original.mean(),
        original.median(),
        original.std(),
        original.skew(),
        original.min(),
        original.max()
    ],
    'After Log Transformation': [
        log_transformed.mean(),
        log_transformed.median(),
        log_transformed.std(),
        log_transformed.skew(),
        log_transformed.min(),
        log_transformed.max()
    ]
}, index=[
    'Mean',
    'Median',
    'Standard Deviation',
    'Skewness',
    'Minimum',
    'Maximum'
])

print("Distribution Comparison:")
print(comparison.round(2))
```

### Observation

The distribution of `lead_time` was compared before and after Log Transformation.

The transformed data shows a reduced effect of large values and may have lower skewness compared with the original data. This demonstrates how transformation can make a skewed feature more suitable for analysis and machine learning.
