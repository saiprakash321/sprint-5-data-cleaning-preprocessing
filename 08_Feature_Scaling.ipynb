# Module 8 -  Feature Scaling

## 1. Why Feature Scaling?

Feature scaling is the process of bringing numerical features to a similar scale.

In the hotel bookings dataset, numerical features such as `lead_time`, `adults`, `stays_in_week_nights`, and `adr` have different ranges.

Without scaling, features with larger numerical values can have a greater influence on algorithms that are sensitive to feature magnitude.

Feature scaling can improve the performance and convergence of many machine learning algorithms.

```python
import pandas as pd
import numpy as np

# Load the raw dataset
df = pd.read_csv(r"C:\Users\HP\sprint-5-data-cleaning-preprocessing\data\hotel_bookings.csv")

# Display the first five records
df.head()
```

```python
numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

print("Feature Ranges Before Scaling:")

for column in numeric_columns:
    print(
        f"{column}: "
        f"Min = {df[column].min()}, "
        f"Max = {df[column].max()}"
    )
```

### Observation

The selected numerical features have different ranges. Feature scaling can bring these features to a comparable scale and prevent scale-sensitive algorithms from being dominated by features with larger values.

## 2. Scale-Sensitive Algorithms

Scale-sensitive algorithms are machine learning algorithms whose performance can be affected by the numerical scale of input features.

These algorithms use distances, gradients, or magnitudes in their calculations.

Examples include:

- K-Nearest Neighbors (KNN)
- Support Vector Machine (SVM)
- Logistic Regression
- K-Means Clustering
- Principal Component Analysis (PCA)
- Neural Networks

For these algorithms, feature scaling is generally important.

```python
scale_sensitive_algorithms = [
    'K-Nearest Neighbors (KNN)',
    'Support Vector Machine (SVM)',
    'Logistic Regression',
    'K-Means Clustering',
    'Principal Component Analysis (PCA)',
    'Neural Networks'
]

print("Scale-Sensitive Algorithms:")

for algorithm in scale_sensitive_algorithms:
    print("-", algorithm)
```

### Observation

The listed algorithms can be influenced by differences in feature scale. Scaling numerical features helps these algorithms treat features more comparably and can improve their performance.

## 3. Scale-Insensitive Algorithms

Scale-insensitive algorithms are machine learning algorithms whose performance is generally not affected by the numerical scale of input features.

These algorithms usually make decisions based on conditions, splits, or categories rather than distances or feature magnitudes.

Examples include:

- Decision Trees
- Random Forest
- Gradient Boosting
- XGBoost
- LightGBM

For these algorithms, feature scaling is generally not required.

```python
scale_insensitive_algorithms = [
    'Decision Trees',
    'Random Forest',
    'Gradient Boosting',
    'XGBoost',
    'LightGBM'
]

print("Scale-Insensitive Algorithms:")

for algorithm in scale_insensitive_algorithms:
    print("-", algorithm)
```

### Observation

The listed algorithms are generally not affected by differences in feature scale. Therefore, feature scaling is usually not required when using these algorithms.

## 4. Normalization

Normalization is a feature scaling technique that transforms numerical values to a specific range, usually between 0 and 1.

It is useful when features have different ranges and the machine learning algorithm is sensitive to feature magnitude.

Normalization is commonly used with distance-based algorithms such as KNN and K-Means.

A common normalization method is Min-Max Scaling.

```python
from sklearn.preprocessing import MinMaxScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(df[numeric_columns])

normalized_df = pd.DataFrame(
    normalized_data,
    columns=numeric_columns,
    index=df.index
)

print("Normalized Data:")
print(normalized_df.head())
```

### Observation

The selected numerical features were normalized to a common range between 0 and 1. This reduces the effect of differences in feature magnitude and is useful for scale-sensitive algorithms.

## 5. Standardization

Standardization is a feature scaling technique that transforms numerical features so that they have a mean of approximately 0 and a standard deviation of approximately 1.

The standardized value is calculated using:

Z = (X - Mean) / Standard Deviation

Standardization is commonly used for algorithms such as Logistic Regression, SVM, KNN, and PCA.

It is generally useful when features have different scales and the data is not restricted to a specific range.

```python
from sklearn.preprocessing import StandardScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = StandardScaler()

standardized_data = scaler.fit_transform(df[numeric_columns])

standardized_df = pd.DataFrame(
    standardized_data,
    columns=numeric_columns,
    index=df.index
)

print("Standardized Data:")
print(standardized_df.head())
```

### Observation

The selected numerical features were standardized to have a mean close to 0 and a standard deviation close to 1. This makes the features comparable in scale for algorithms that are sensitive to feature magnitude.

## 6. Min-Max Scaling

Min-Max Scaling is a feature scaling technique that transforms numerical values to a fixed range, usually between 0 and 1.

The formula is:

X_scaled = (X - X_min) / (X_max - X_min)

It preserves the relative relationships between the values while bringing all features to the same scale.

Min-Max Scaling is useful for scale-sensitive algorithms such as KNN, K-Means, and Neural Networks.

```python
from sklearn.preprocessing import MinMaxScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df[numeric_columns])

minmax_df = pd.DataFrame(
    scaled_data,
    columns=numeric_columns,
    index=df.index
)

print("Min-Max Scaled Data:")
print(minmax_df.head())

print("\nMinimum values:")
print(minmax_df.min())

print("\nMaximum values:")
print(minmax_df.max())
```

### Observation

The selected numerical features were transformed to a range between 0 and 1 using Min-Max Scaling. This makes the features comparable and reduces the effect of differences in their original ranges.

## 7. StandardScaler

`StandardScaler` is a preprocessing technique from scikit-learn used to standardize numerical features.

It transforms the data so that each feature has a mean of approximately 0 and a standard deviation of approximately 1.

The transformation is based on:

Z = (X - Mean) / Standard Deviation

StandardScaler is commonly used for algorithms such as KNN, SVM, Logistic Regression, and PCA.

```python
from sklearn.preprocessing import StandardScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df[numeric_columns])

standardized_df = pd.DataFrame(
    scaled_data,
    columns=numeric_columns,
    index=df.index
)

print("StandardScaler Output:")
print(standardized_df.head())

print("\nMean after scaling:")
print(standardized_df.mean().round(2))

print("\nStandard deviation after scaling:")
print(standardized_df.std().round(2))
```

### Observation

The selected numerical features were standardized using `StandardScaler`. The resulting features have a mean close to 0 and a standard deviation close to 1.

## 8. MinMaxScaler

`MinMaxScaler` is a preprocessing technique from scikit-learn used to scale numerical features to a specified range, usually between 0 and 1.

It uses the minimum and maximum values of each feature to perform the transformation.

MinMaxScaler is useful when the features need to be brought to a common range, especially for scale-sensitive algorithms such as KNN, K-Means, and Neural Networks.

```python
from sklearn.preprocessing import MinMaxScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df[numeric_columns])

minmax_df = pd.DataFrame(
    scaled_data,
    columns=numeric_columns,
    index=df.index
)

print("MinMaxScaler Output:")
print(minmax_df.head())

print("\nMinimum values after scaling:")
print(minmax_df.min().round(2))

print("\nMaximum values after scaling:")
print(minmax_df.max().round(2))
```

### Observation

The selected numerical features were scaled using `MinMaxScaler`. The values were transformed to a range between 0 and 1, making the features comparable in scale.

## 9. RobustScaler

`RobustScaler` is a preprocessing technique from scikit-learn that scales numerical features using the median and interquartile range (IQR).

It is less affected by outliers compared with StandardScaler and MinMaxScaler.

RobustScaler is useful when the dataset contains extreme values or outliers.

The scaling is based on:

X_scaled = (X - Median) / IQR

For the hotel bookings dataset, `adr` can contain extreme values, so RobustScaler can be useful for scaling this feature.

```python
from sklearn.preprocessing import RobustScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = RobustScaler()

scaled_data = scaler.fit_transform(df[numeric_columns])

robust_df = pd.DataFrame(
    scaled_data,
    columns=numeric_columns,
    index=df.index
)

print("RobustScaler Output:")
print(robust_df.head())

print("\nMedian after scaling:")
print(robust_df.median().round(2))
```

### Observation

The selected numerical features were scaled using `RobustScaler`. Since it uses the median and IQR, it is less sensitive to extreme values and can be useful for datasets containing outliers.

## 10. MaxAbsScaler

`MaxAbsScaler` is a preprocessing technique from scikit-learn that scales numerical features based on their maximum absolute value.

It transforms each feature so that its values are generally within the range of -1 to 1.

MaxAbsScaler is useful when the data contains zero values or sparse data and we want to preserve the sign of the values.

It does not center the data around zero and is less suitable when strong outliers are present.

```python
from sklearn.preprocessing import MaxAbsScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scaler = MaxAbsScaler()

scaled_data = scaler.fit_transform(df[numeric_columns])

maxabs_df = pd.DataFrame(
    scaled_data,
    columns=numeric_columns,
    index=df.index
)

print("MaxAbsScaler Output:")
print(maxabs_df.head())

print("\nMaximum absolute values after scaling:")
print(maxabs_df.abs().max().round(2))
```

### Observation

The selected numerical features were scaled using `MaxAbsScaler`. The maximum absolute value of each feature was scaled to 1 while preserving the sign of the original values.

## 11. Compare Different Scaling Techniques

Different scaling techniques are suitable for different types of data.

- StandardScaler: Centers data around 0 with a standard deviation of 1.
- MinMaxScaler: Scales values to a fixed range, usually 0 to 1.
- RobustScaler: Uses the median and IQR, making it less sensitive to outliers.
- MaxAbsScaler: Scales values based on the maximum absolute value and preserves the sign of the data.

The choice of scaling technique depends on the data distribution, presence of outliers, and requirements of the machine learning algorithm.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, MaxAbsScaler

numeric_columns = [
    'lead_time',
    'adults',
    'children',
    'stays_in_week_nights',
    'adr'
]

scalers = {
    'StandardScaler': StandardScaler(),
    'MinMaxScaler': MinMaxScaler(),
    'RobustScaler': RobustScaler(),
    'MaxAbsScaler': MaxAbsScaler()
}

comparison = {}

for name, scaler in scalers.items():
    scaled_data = scaler.fit_transform(df[numeric_columns])
    comparison[name] = scaled_data.mean(axis=0)

comparison_df = pd.DataFrame(
    comparison,
    index=numeric_columns
)

print("Scaling Technique Comparison:")
print(comparison_df.round(2))
```

### Observation

Different scaling techniques produce different transformed values because they use different scaling methods. StandardScaler and MinMaxScaler are commonly used for general scaling, while RobustScaler is useful when outliers are present and MaxAbsScaler is useful when preserving the sign of the data is important.

## 12. How Outliers Affect Scaling

Outliers are extreme values that are significantly different from most observations.

Outliers can affect feature scaling differently depending on the scaling technique.

- MinMaxScaler: Highly affected by outliers because it uses minimum and maximum values.
- StandardScaler: Can be affected by outliers because it uses the mean and standard deviation.
- RobustScaler: Less affected by outliers because it uses the median and IQR.
- MaxAbsScaler: Can be affected by extreme values because it uses the maximum absolute value.

Therefore, the presence of outliers should be considered when selecting a scaling technique.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, MaxAbsScaler

column = 'adr'

scalers = {
    'StandardScaler': StandardScaler(),
    'MinMaxScaler': MinMaxScaler(),
    'RobustScaler': RobustScaler(),
    'MaxAbsScaler': MaxAbsScaler()
}

print("Effect of Outliers on Scaling:")

for name, scaler in scalers.items():
    scaled_values = scaler.fit_transform(df[[column]])
    
    print(f"\n{name}:")
    print("Minimum:", scaled_values.min())
    print("Maximum:", scaled_values.max())
```

### Observation

The scaling methods respond differently to extreme `adr` values. MinMaxScaler and MaxAbsScaler can be strongly influenced by extreme values, while RobustScaler is less affected because it uses the median and IQR.

Therefore, RobustScaler can be a suitable choice when significant outliers are present.
